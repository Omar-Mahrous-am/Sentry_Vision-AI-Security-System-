```python
text1="There is fire in Area 3 near to Gate 14 please call Emergency !!"
text2="a Stolen Car detected near to Gate 4  please call Police !!"
```

```python
from gtts import gTTS
from IPython.display import Audio
import os

# Combine the text 4 times for a single repeating audio file
repeated_text = (text1 + " ") * 4

tts = gTTS(text=repeated_text, lang='en')
tts.save("emergency_alert.mp3")

# This will play the combined message in one go
Audio("emergency_alert.mp3", autoplay=True)
```
